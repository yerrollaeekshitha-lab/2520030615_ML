```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
#----------------------------------------------
#1.load data
#----------------------------------------------
df=pd.read_csv("/content/housing.csv")
target_col = "median_house_value"
#Drop the ID column and CGPA_Tier (a label derived direcctly from CGPA,
#so keeping it would leak the target) before selecting numeric features


drop_cols = [target_col]

feature_df = df.drop(columns=drop_cols).select_dtypes(include=[np.number])
#Some columns (workshops,AptitudeTestScore,SoftSkillRating,
#CodingTestScore, MockInterviewScore) have missing values -
#impute with the column median so gradient descent doesn't blow up on NaNs
feature_df=feature_df.fillna(feature_df.median())
X=feature_df.values
y=df[target_col].values.reshape(-1, 1)
```

```python
print(df.columns.tolist())
```

```python

y = df[target_col]
```

```python

features = [
    "longitude",
    "latitude",
    "housing_median_age",
    "total_rooms",
    "total_bedrooms",
    "population",
    "households",
    "median_income",
    "ocean_proximity"
]

X = df[features].copy()
```

```python
categorical_features = [
    "ocean_proximity"
]
```

```python
numerical_features = [
    "longitude",
    "latitude",
    "housing_median_age",
    "total_rooms",
    "total_bedrooms",
    "population",
    "households",
    "median_income"
]
```

```python
for col in numerical_features:
    X[col] = X[col].fillna(X[col].median())

for col in categorical_features:
    X[col] = X[col].fillna(X[col].mode()[0])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first"
            ),
            categorical_features
        )
    ]
)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", LinearRegression())
    ]
)
```

```python
linear_model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
```

```python
y_pred = linear_model.predict(X_test)

print("First 10 Predictions:")
print(y_pred[:10])
```
